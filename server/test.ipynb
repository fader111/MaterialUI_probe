```python
# генератор qr кодов
import qrcode

# Ввод данных
data = ''' 
        # генератор qr кодов
        import qrcode

        # Ввод данных
        data = "Да здравствует созданный волей народов"  # Можно заменить на любой текст или ссылку

        # Создание QR-кода
        qr = qrcode.QRCode(
            version=1,  # Размер QR-кода (от 1 до 40)
            error_correction=qrcode.constants.ERROR_CORRECT_L,  # Уровень коррекции ошибок
            box_size=10,  # Размер каждой коробки QR-кода
            border=4,  # Ширина границы (в коробках)
        )
        qr.add_data(data)
        qr.make(fit=True)

        # Генерация изображения
        img = qr.make_image(fill_color="black", back_color="white")

        # Сохранение изображения
        img.save("qr_code.png")
    '''
# Создание QR-кода
qr = qrcode.QRCode(
    version=1,  # Размер QR-кода (от 1 до 40)
    error_correction=qrcode.constants.ERROR_CORRECT_L,  # Уровень коррекции ошибок
    box_size=10,  # Размер каждой коробки QR-кода
    border=4,  # Ширина границы (в коробках)
)
qr.add_data(data)
qr.make(fit=True)

# Генерация изображения
img = qr.make_image(fill_color="black", back_color="white")

# Сохранение изображения
img.save("qr_code.png")
```
