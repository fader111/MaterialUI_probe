```python
def get_supernum_teeth(oas_path):
    doc = PyOrthoDoc()
    # oas_path = 'D:/projectsd/MaterialsUI_probe/server/00000000.oas'
    doc.OpenCase(oas_path)
    tp = doc.GetActiveTreatmentPlan()
    supernum_teeth = {}
    for jawType in JawType:
        jaw = tp.GetJaw(jawType)
        for tooth in jaw.getTeeth():
            clID = tooth.getClinicalID()
            # print(clID, isUpper(clID), isLower(clID), type(clID), jawType, type(jawType))
            if clID > 50: 
                supernum_teeth[clID] = oas_path.split(os.sep)[-1]
    return supernum_teeth

# ищем кейсы с суперномерарными зубами
oas_folder = r'D:\awsCollectedData3'
oas_files = [os.path.join(oas_folder, f) for f in os.listdir(oas_folder) if f.endswith('.oas')][:1]
for idx, oas_path in enumerate(oas_files):
    if idx%10 == 0:
        print(f"Processed {idx} files")
    supernum = get_supernum_teeth(oas_path)
    if not supernum:
        continue
    print(f"File: {oas_path}, Supernumerary Teeth: {list(supernum.keys())}")
```
