```python
import os
import subprocess
import json

EXE_PATH = r"E:\\WebGLServer\\orthoplatform\\Build\\windows-msbuild-cl\\Bin\\OASDatabase\\Release\\OASDatabase.exe"
# Get absolute path to the current notebook directory
NOTEBOOK_DIR = os.path.dirname(os.path.abspath("__file__")) if '__file__' in globals() else os.getcwd()
TEMPLATE_DIR = os.path.abspath(os.path.join(NOTEBOOK_DIR, "templates"))
PUBLIC_DIR = os.path.abspath(os.path.join(NOTEBOOK_DIR, "public"))

# Check if the template directory exists
if not os.path.isdir(TEMPLATE_DIR):
    raise FileNotFoundError(f"Template directory not found: {TEMPLATE_DIR}")

# List all .oas files in the template directory
oas_files = [f for f in os.listdir(TEMPLATE_DIR) if f.endswith('.oas')]
print(f"Found {len(oas_files)} .oas files in {TEMPLATE_DIR}")

for oas_file in oas_files:
    oas_path = os.path.join(TEMPLATE_DIR, oas_file)
    orthoDataFilePath = os.path.join(TEMPLATE_DIR, f"{os.path.splitext(oas_file)[0]}_orthoData.json")
    job = [
        EXE_PATH,
        "ExportStagingData",
        oas_path,
        orthoDataFilePath
    ]
    print(f"Processing {oas_file} ...")
    result = subprocess.run(job, capture_output=True, text=True)
    if result.returncode != 0:
        print(f"ERROR: Subprocess failed for {oas_file}: {result.stderr}")
        continue
    if not os.path.isfile(orthoDataFilePath):
        print(f"ERROR: orthoData.json not found for {oas_file}")
        continue
    print(f"orthoData.json created: {orthoDataFilePath}")
```
